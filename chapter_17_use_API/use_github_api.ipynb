```python
import requests
import urllib3

# 1. 屏蔽底层的 SSL 安全警告（不然控制台会一直报红字）
urllib3.disable_warnings()

# 执行API调用，并查看反应
url = "https://api.github.com/search/repositories"
url += "?q=language:python+sort:stars:>10000"

headers = {"Accept": "application/vnd.github.v3+json"}
# headers 的全称是 HTTP Headers（请求头）
# 含义：设置请求头（HTTP Headers）。
# 解释：Accept 这个参数告诉 GitHub：“我是一个 v3+json 版本的消费者，请把数据以 JSON 格式 返回给我。”
# Accept是固定的键，后面值是随意的。application/vnd.github.v3+json是github规定的格式。
# 这就像是点外卖时告诉商家：“我的口味是要标准版，不要改配方。”

# 固定的 Key	                            它的含义	                                 你能改的是
# Accept	                      告诉服务器：“我期望收到什么格式”	                后面的 Value 可以改
# User-Agent	                  告诉服务器：“你是谁（浏览器/爬虫）”	            后面的 Value 可以改（比如你的项目名）
# Authorization	                  告诉服务器：“我的身份凭证（登录令牌）”	        后面的 Value 是真正的 API 密钥，不能随便

r = requests.get(url, headers=headers, verify=False)
# my_website = "https://api.github.com/search/repositories"，则r=requests.get(url=my_website)
# 执行 HTTP 请求，拿到结果。
# requests.get()：用 GET 方法（获取数据）请求服务器。
# url：你要访问的具体网址。
# headers：带上你刚才设置的请求头。
# verify=False：跳过 HTTPS 证书验证（只用于学习开发，躲避国内网络环境的 SSL 拦截）。
# 结果：拿到一个 Response 对象，赋值给变量 r。


# 三步走”顺序不能乱：
# requests.get()（先去拿包裹）
# r.status_code（先看一眼状态码）
# r.json()（再打开包裹看数据）
# 如果你连包裹都没拿到（没有 r），你拿什么去拆？ 所以 r.json() 必须“寄居”在 r 这个对象身上，而 r 必须来源于 requests.get()。你的思考逻辑非常严谨！

print(f"Status code:{r.status_code}")
# r = requests.get(url) 返回的是一个 Response 对象。
# 在这个对象内部，requests 库预设了很多数据。其中一个是状态码（比如 200、404），requests 把它命名为 status_code。
#  200：正常        404：找不到        403：无权访问      429：请求频繁      500：服务器崩溃       503：服务器休息
# 这行是“安全检查”，确定你的请求有没有被服务器接纳
# r.status_code: r 的属性。requests模块的固定用法

# 将响应转换为字典：
response_dict = r.json()
# response: 回应,回复。回信” 或 “答复”。专用的翻译： 返回值
# 含义：把服务器返回的 JSON 仓库对象，转换成 Python 字典。
# 解释：服务器返回的其实是一长串纯文本（JSON 格式）。
# r.json() 是 requests 库提供的一个魔法函数，它像 json.loads() 一样，把这串纯文本变成 Python 能操作的大字典。

# 处理结果：
print(response_dict.keys())
# response_dict.keys()，返回的是一个可视对象
# 打印字典的键。含义：看看这个大字典里到底有什么。
# 解释：这是最经典的数据探索步骤！因为你不确定 GitHub 到底返回了什么，你需要先看它的形状（有哪些键）。
# 以上是你正在给 GitHub 服务器发一张“订单”（请求），服务器把数据以 JSON 格式打包返回，而你在小心翼翼地拆开这个包裹，看看里面都装了些什么。
```

```python
# 处理响应字典
# 生成一些概述这些信息的输出
import requests
import urllib3

# 1. 屏蔽底层的 SSL 安全警告（不然控制台会一直报红字）
urllib3.disable_warnings()

# 执行API调用，并查看反应
url = "https://api.github.com/search/repositories"
url += "?q=language:python+sort=stars&stars:>10000"

# sort()方法：python 数字从小到大，字母按as||码的顺序从前往后排。
# sort()方法直接把原来的内容顺序改变。只能用在list上。
# sorted()函数，不改变原来的列表，而是生成一个新的排序过的列表。list,set,dict,teple都可以用。

headers = {"Accept": "application/vnd.github.v3+json"}
r = requests.get(url, headers=headers, verify=False)
# “r 是一个 Response 对象，它里面装的是 JSON 格式的文本字符串。你用 r.json() 把它转化成了 Python 能看懂的字典。”
# 因为 r（Response）本身不是数据，它是一个容器（快递盒子），里面装了一个 JSON 格式的纯文本字符串。
# 如果你直接打印 r，你只会看到 <Response [200]>，根本看不到数据。
# 必须先有r=requests.get()，后面才能r.json().
print(f"Status code:{r.status_code}")

# 将响应转换为字典：
response_dict = r.json()
# 对象r，是一个大的容器。接收到回信的字典，被转换为python认识的字典。
print(f"Total repositories: {response_dict['total_count']}")
# 打印一个字典的键，dict['key'],=value.得到对应的键值。在 Python 里，方括号 [] 代表的是“索引”或“键值查找”。
# repository: 仓库
# 查看仓库的总数
print(f"Complete results: {not response_dict['incomplete_results']}")
# imcomplete: 不完整的，未完成的。
# 这次搜索已完成。

# 搜索有关仓库的信息
repo_dicts = response_dict["items"]
# 收到的回复中所有有关'items'的列表，即：仓库名
# repo:报告report.
print(f"Repositories resulted: {len(repo_dicts)}")

# 研究第一个仓库：
repo_dict = repo_dicts[0]
# 取repo_dicts中的第一个值，并赋值给repo_dict.这个repo_dict也是一个字典。
print(f"\nKeys:{len(repo_dict)}")
# 第一个仓库中，有82个键值对
for key in sorted(repo_dict.keys()):
    # 对字典repo_dict的所有的键，按照字母顺序排序，变成一个列表，然后遍历。
    # 因为 sorted() 的核心功能是“排序并返回一个新的有序列表”，它不需要修改你原来的数据。所以可以sorted(tuple,list,set)得到一个新列表。
    # 不能sorted(number,bool）等，普通的字典也不行，只能排字典中的key。
    # sort()：是一个方法（Method），从高到低排列。它只能用在列表上。它必须修改原列表才能生效。因为元组不可变，所以元组没有 sort()。
    # sorted()：是一个全局函数（Function），它可以用在任何可迭代对象上。它不修改原件，而是重新返回一个列表。所以无论你给它是列表、元组还是集合，它都能排！
    print(key)

print(response_dict.keys())
```

```python
# 使用plotly可视化仓库
# 设置图形样式
# 添加可以单击 的 链接

import requests
import urllib3
import plotly.express as px

# 1. 屏蔽底层的 SSL 安全警告（不然控制台会一直报红字）
urllib3.disable_warnings()

# 执行API调用，并查看反应
url = "https://api.github.com/search/repositories"
url += "?q=language:python+stars:>10000&sort=stars"

headers = {"Accept": "application/vnd.github.v3+json"}
r = requests.get(url, headers=headers, verify=False)

print(f"Status code:{r.status_code}")

# 将响应转换为字典：
response_dict = r.json()
print(f"Complete results: {not response_dict['incomplete_results']}")


# 搜索有关仓库的信息
repo_dicts = response_dict["items"]
repo_names, stars, hover_texts, repo_links = [], [], [], []
for repo_dict in repo_dicts:
    repo_names.append(repo_dict["name"])
    stars.append(repo_dict["stargazers_count"])

    # 将仓库名转为链接
    repo_name = repo_dict["name"]
    repo_url = repo_dict["html_url"]
    repo_link = f"<a href='{repo_url}'>{repo_name}</a>"
    repo_links.append(repo_link)

    # 创建悬停文本
    owner = repo_dict["owner"]["login"]
    description = repo_dict["description"]
    hover_text = f"{owner}<br />{description}"
    # plotly 生成的图表，底层不是普通的图片，而是一个 HTML 网页（用 SVG 和 JavaScript 写的），所以用<br />html语言写。
    hover_texts.append(hover_text)

title = "Most-Starred Python Projects on GitHub"
labels = {"x": "Repository", "y": "Stars"}
fig = px.bar(x=repo_links, y=stars, title=title, labels=labels, hover_name=hover_texts)
# 假设你的图已经生成了 fig
fig.update_traces(
    hoverlabel=dict(
        bgcolor="black",  # 悬停框背景：纯黑
        font_color="white",  # 悬停框文字：纯白
        bordercolor="white",  # 悬停框边框：纯白（可选）
    )
)
fig.show()

fig.update_layout(
    title_font_size=28, xaxis_title_font_size=20, yaxis_title_font_size=20
)
# fig.update_layout: 在不改变图中数据的情况下，修改图表的“外观、排版和布局”。

fig.update_traces(marker_color="SteelBlue", marker_opacity=0.6)
# fig.update_traces()：设置修改图形上的一系列标记。marker_opacity:不透明度1.0：完全不透明。

fig.show()
```
