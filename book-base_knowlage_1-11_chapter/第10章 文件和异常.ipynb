```python
# 异常：
# 当你预感到，或许代码会出错误的时候，把它写入一个try-excetp代码块，并预测会是什么错误。
# 运行代码如果真的出现了预测的错误，python将把错误类型告诉你，而且并不影响代码继续运行，不会崩溃。
# 可以预测多个错误类型。
# 如果出现的错误不在预测的错误类型中，程序将崩溃。
try:
    print("a" / "b")

except ZeroDivisionError:
    print("0 不能当除数")

except NameError:
    print("名称都搞错啦")

except TypeError:
    print("字符串不能运算")
```

```python
# 来一个只运算除法的程序：

print("输入两个数字，做个除法。")
print("输入'q',就退出")


while True:
    first_number = input("\nfirst_number is :")
    if first_number == "quit":
        break
    second_number = input("\nsecond_number is :")
    if second_number == "quit":
        break

    try:
        # try-except语句的缩进要在while内部，因为是测试其代码。
        answer = int(first_number) / int(second_number)
        # print(answer)
    except TypeError:
        print("请输入数字")

    except ValueError:
        print("输入非法")

    else:
        print(answer)
    # try-except-else 是官方推荐的模式。
    # 使用 else 更清晰、更安全。它明确区分了“可能出错的代码”和“正常执行的代码”，是更好的实践。


#     🔍 具体对比
# 对比维度	TypeError	                                       ValueError
# 关注点	操作的数据类型是否正确	                           数据的值是否合法/有意义
# 发生时机	类型不匹配时	                                   类型正确，但值本身不合法时
# 常见场景	对数字和字符串做加法、调用不存在的方法	           int("abc")、超出范围的数字
```

```python
# 用json.loads()函数读取json格式的文档内容：


from pathlib import Path
import json

num = [2, 3, 5, 7, 11, 13]
path = Path("number.json")
contents = json.dumps(num)
# 此时，json.dumps()返回的是一个字符串。
# json.dumps(num) 返回的是 JSON 格式的字符串，打印出来看起来像列表，但实际上是字符串(json格式的字符串)

#  dump:字面意思与编程含义
# 字面意思（日常英语）： 倾倒、倒出、卸货（比如把垃圾倒掉，或者把卡车里的货卸下来）。
# 编程含义： 把内存中的数据（比如 Python 的字典或列表），“倾倒” 到一个能够永久保存的地方（通常是硬盘上的文件），
# 或者“倾倒” 成一种能传输的格式（比如 JSON 字符串）。

# 2. json.dumps() 到底在做什么？

# 你之前用了 json.loads()（把字符串变成字典）。
# 现在 json.dumps() 是反向操作：把 Python 字典变成 JSON 字符串。

# dumps = dump + string（转储成字符串）。
# 作用： 把内存里的 Python 对象，转化成可以写入文件、或者通过网络传输的 JSON 格式文本。


path.write_text(contents)
# write_text()方法得以把字符串写入‘number.json'文件。

numbers = json.loads(contents)
# json.loads() 只能解析 JSON 格式的字符串，不能直接解析列表、字典、集合。
# 将json格式的文本内容读出。
print(numbers)
print(path.read_text())
print(contents)
print(json.dumps(num))

# read_text() 和 json.loads() 的作用完全不同：

# read_text() 可以读字典、集合吗？
# read_text() 不关心内容是什么，它只读取文件的原始文本。


# 方法	                     作用	                                             返回值类型
# path.read_text()	         读取文件的原始文本内容	                             str（字符串）
# json.loads()	             把 JSON 格式的字符串转换成 Python 对象	             list / dict / int / str 等

# 🔍 对比

# from pathlib import Path
# import json

# path = Path("number.json")
# contents = json.dumps([2, 3, 5, 7, 11, 13])
# path.write_text(contents)

# 1. read_text()：读取原始文本（字符串）
# raw_text = path.read_text()
# print(raw_text)          # [2, 3, 5, 7, 11, 13]   原文是json格式的，表面看是列表，其实是字符串。
# print(type(raw_text))    # <class 'str'>

# 2. json.loads()：把 JSON 字符串转换成 Python 对象
# numbers = json.loads(raw_text)
# print(numbers)           # [2, 3, 5, 7, 11, 13]
# print(type(numbers))     # <class 'list'>
# 📌 它们的分工
# text
# 文件（number.json） → read_text() → 字符串（JSON 格式）
#         ↓
# 字符串（JSON 格式） → json.loads() → Python 对象（列表/字典）


# 📌 什么时候用哪个？
# 场景	                                                 方法
# 只看文件里写了什么（不解析）                      	read_text()
# 需要把 JSON 数据变成 Python 对象来操作	            json.loads()
# 📌 一句话总结
# read_text() 读的是“原文”，返回字符串；原文是啥格式，读取的就是啥格式，但是都读成 字符串。
# json.loads() 把 JSON 格式的字符串“解析”成 Python 对象。
# 两者配合使用：先 read_text() 读文件，再用 json.loads() 解析内容。


# read_text() 只读原文（以字符串的形式），不管文件里写的是什么字典列表集合。
# json.loads() 必须接收字符串，
# 而 json.load() 可以直接读取 JSON（字符串） 文件并解析成 Python 对象。
# json.dumps()把python格式的字符串转为json的格式的字符串
```

```python
import json
from pathlib import Path


def get_stored_username(path):
    """如果文件存在，读取并返回用户名；否则返回 None"""
    if path.exists():
        contents = path.read_text()
        return json.loads(contents)
    return None


def check_username(stored_username):
    """验证输入的名字是否匹配存储的名字"""
    name = input("enter your name : ")
    if name == stored_username:
        print(f"here you are {name}")
        return True
    else:
        print("名单没有你")
        return False


def create_new_user(path):
    """创建新用户，写入文件"""
    name = input("enter your name : ")
    path.write_text(json.dumps(name))
    print(f"we will remember you {name}")


def greet_user():
    """主控函数：协调整个流程"""
    path = Path("username_today.json")
    stored_username = get_stored_username(path)

    if stored_username:
        check_username(stored_username)
    else:
        create_new_user(path)


if __name__ == "__main__":
    greet_user()
```

```python
# 变量名的起名窍门：


# 📌 一套通用的变量命名模板（存下来，直接用）

# 场景	                            变量名	                                         说明
# 文件路径	                        path	                                         永远叫 path
# 用户输入	                        user_input	                                     或 name、age 等
# 原始数据（Python 对象）	        data	                                         最终要用的数据
# JSON 字符串	                    json_str	                                     转成 JSON 格式的字符串
# 读取的内容	                    content	                                         从文件读出来的原始文本
# 写入的内容	                    content	                                         同上，读写都用 content
# 计数器/索引	                    i、count、index	                                 循环专用


# 📌 为什么这样命名不容易乱？
# path 永远是路径，不会变。

# user_input 永远是用户输入，不会变。

# json_str 永远是 JSON 字符串，不会变。

# data 永远是你最终要用的数据。

# 你不需要为每个步骤发明新名字，只需要知道当前数据处于哪个阶段。

# 📌 一句话总结
# 变量名混乱是因为你把“数据的状态”写进了名字里（如 convert、read_out）。

# 正确的做法是用“数据的形态”命名（如 json_str、data），不管它在哪个阶段，名字不变。

# 这样代码更短、更清晰、更不容易犯错。
```
