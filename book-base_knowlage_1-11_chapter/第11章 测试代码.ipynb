```python
# 什么情况下要有return?有没有return 有啥区别，有啥后果？什么时候必须有return ?
# 如果后续代码需要用到这个结果，就return ,如果不需要，就可以不用return.

# 补充一点：修改外部对象的函数通常不需要 return，纯计算函数必须有 return。

# 📌 唯一要注意的：参数类型
# 如果你的函数是修改外部对象（比如列表、字典、类实例），通常不需要 return，因为修改会影响原对象：
# 但如果函数是纯计算（不修改传入参数），通常需要 return：

#  一、`return` 到底在干什么？

# `return` 的作用是：**把函数内部的值“送出去”**，送给调用函数的那行代码。


def get_formatted_name(first, last):
    full_name = f"{first}.{last}"
    return full_name  # ← 把 full_name 送出去


result = get_formatted_name("mikle", "bodun")
# get_formatted_name() 被替换成 "Mikle.Bodun"
# result = "Mikle.Bodun"


# 没有 `return`：
def get_formatted_name(first, last):
    full_name = f"{first}.{last}"
    # 没有 return


result = get_formatted_name("mikle", "bodun")
print(result)  # None
# 函数内部计算了 `full_name`，但**没有送出来**，所以调用者拿到的只有 `None`。


## 二、什么情况下必须有 `return`？

# | 场景                                          | 是否必须有 `return`                      | 说明 |

# | 函数需要**返回一个值**给调用者                | ✅ **必须有**                           | 比如：计算结果、处理后的数据 |
# | 函数只是**打印**或**修改外部状态**            | ❌ **不需要**                            | 比如：`print()`、`list.append()` |
# | 函数需要**提前结束**（遇到错误）              | ✅ **可以有**（但返回 `None`）          | 比如：`if error: return` |
# | 函数需要**返回多个值**                        | ✅ **必须有**                           | 用元组返回：`return a, b, c` |


### ✅ 必须有 `return` 的典型场景
def add(a, b):
    return a + b  # ✅ 必须返回结果


def get_user_name(user_id):
    return users[user_id]  # ✅ 必须返回查询结果


def calculate_average(scores):
    total = sum(scores)
    return total / len(scores)  # ✅ 必须返回计算结果


### ❌ 不需要 `return` 的典型场景
def greet(name):
    print(f"Hello, {name}")  # ❌ 不需要 return，只打印


def update_user(user):
    user["age"] = 18  # ❌ 不需要 return，直接修改传入的对象


## 三、有 `return` 和没有 `return` 的后果

# | 对比项                                   | **有 `return`**                                     | **没有 `return`** |

# | 调用结果                                 | 得到 `return` 后面的值                              | 得到 `None` |
# | 能否赋值给变量                           | ✅ 可以 `result = func()`                          | ⚠️ 可以，但 `result = None` |
# | 能否用在表达式中                         | ✅ 可以 `print(func())`                            | ⚠️ 可以，但打印 `None` |
# | 函数是否有“输出”                         | ✅ 有明确输出                                      | ❌ 没有输出，只有副作用 |


# | 测试时 | 可以直接断言返回值 | 只能检查外部状态或捕获输出 |
### 🧪 有 `return` 的测试（pytest）


# 有 return 的函数
def get_formatted_name(first, last):
    return f"{first}.{last}".title()


# 测试时直接断言返回值
def test_get_formatted_name():
    assert get_formatted_name("mikle", "bodun") == "Mikle.Bodun"


### ❌ 没有 `return` 的测试
# 没有 return 的函数
def print_formatted_name(first, last):
    print(f"{first}.{last}".title())


# 测试时不能直接断言，需要捕获 print 输出
def test_print_formatted_name(capsys):
    print_formatted_name("mikle", "bodun")
    captured = capsys.readouterr()
    assert captured.out == "Mikle.Bodun\n"


## 四、`return` 的默认行为

# **任何函数都有返回值。** 即使你不写 `return`，Python 也会在函数末尾隐式地执行 `return None`。


def func_without_return():
    x = 1 + 2
    # 隐含：return None


result = func_without_return()
print(result)  # None


## 五、`return` 还会立即结束函数
def divide(a, b):
    if b == 0:
        print("除数不能为0")
        return  # ← 提前结束函数
    return a / b  # ← 只有 b != 0 时才执行


## 📌 一句话总结

# > **需要“输出结果”时必须有 `return`，否则函数返回 `None`。
# `return` 不仅决定了返回值，还决定了调用者能否拿到计算结果。
# 测试时，有 `return` 的函数更容易测试。**
```

```python
# 写一个需要测试的类


class AnonymousSurvey:
    """收集匿名调查问卷的答案"""

    # survey:调查问卷

    def __init__(self, question):
        """存储一个问题，并未存储答案做准备"""
        self.question = question
        self.response = []
        # response:回答，回复

    def show_question(self):
        """显示调查问卷"""
        print(self.question)

    def store_response(self, new_response):
        """存储单份调查答案"""
        self.response.append(new_response)

    def show_results(self):
        """显示收集到的所有答案"""
        print("Survey results: ")
        for response in self.response:
            print(f" - {response}")


# import sys
# from pathlib import Path

# sys.path.append(
#     str(Path("D:/python_learning/python入门到实践by_jupyter/pytest_test_dir"))
# )
# from survey import AnonymousSurvey

"""定义一个问题，并创建一个表示调查的AnonymouseSurvey对象"""
question = "what language did you first learn to speak?"
language_survey = AnonymousSurvey(question)

# 显示问题并存储答案：
language_survey.show_question()
print("enter 'q' at any time to quit.\n")
while True:
    response = input("Language is : ")
    if response == "q":
        break
    language_survey.store_response(response)

# 显示调查结果
print("\nThank you to everyone who participated in the survey!")
language_survey.show_results()
```
