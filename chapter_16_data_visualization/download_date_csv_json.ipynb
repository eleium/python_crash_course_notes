```python
# 打印文件头及其位置：
# CSV文件：将数据组织成一系列的以逗号分隔的值（commaseparated values,CSV)并写入文件，叫CSV文件
# comma: 逗号   separate: 分隔  value:值       逗号分隔值

from pathlib import Path
import csv

path = Path("weather_data/sitka_weather_07-2021_simple.csv")
# 1. 指定文件路径

lines = path.read_text().splitlines()
# 2. path.read_text() 把整个文件的内容读取为一个长字符串
# 3. splitlines() 把这个长字符串按换行符切成一个列表（每行是一个元素）

reader = csv.reader(lines)
# 4. csv.reader(lines) 创建一个解析器对象，这个对象能一行一行地解析 CSV 数据

header_row = next(reader)
# 5. next(reader) 把解析器的“第一行”取出来（通常第一行是表头）
# 6. 单独取第一行的目的是：方便了解列名，同时让后续的循环从实际数据（第二行）开始

print(header_row)

for index, column_header in enumerate(header_row):
    # enumerate():英文原意是 “列举、枚举”。在编程里，它特指 “为列表中的元素按顺序编号”。。
    # 它的核心作用是：在遍历一个可迭代对象（如列表、元组、字符串）时，同时获取“元素的索引（下标）”和“元素的值”。
    # 如果你想看它里面装了什么，必须把它转化成列表：是一个迭代器，每次用的时候，取一个值出来。打印不出来具体的值，要用list转化后能打印。
    # print(list(enumerate(header_row)))
    #  输出: [(0, 'Day'), (1, 'High'), (2, 'Low'), (3, 'Avg')]
    # 看到重点了吗？ 当你把它转成列表时，它其实是一个包含多个“元组”的列表。每个元组都是 (索引, 值) 的配对。
    print(index, column_header)
```

```python
# 提取并读取数据
# 打印文件头及其位置：
# CSV文件：将数据组织成一系列的以逗号分隔的值（commaseparated values,CSV)并写入文件，叫CSV文件
# comma: 逗号   separate: 分隔  value:值       逗号分隔值

from pathlib import Path
import csv

path = Path("weather_data/sitka_weather_07-2021_simple.csv")
lines = path.read_text().splitlines()
reader = csv.reader(lines)
# csv.reader(lines) 创建一个解析器对象，这个对象能一行一行地解析 CSV 数据
header_row = next(reader)
# row: 行  column:列
# 返回文件头部的一个列表,把 CSV 文件里的第一行（表头行）拿出来，存起来

# next()与pop():
# next() 像“翻书”，翻走一页，书还在那里，只是你的目光落到下一页了。
# pop() 像“撕书”，撕掉一页，书就永远少了一页。
# 如果你不想把书翻乱，可以先把整本书复印一份（list(reader)），然后在复印件上随意操作。

# for index, column_header in enumerate(header_row):
#     # 将该列表(也就是第一行)以 下标 元素值 的方式循环
#     print(index, column_header)

# 读取最高温度：
highs = []
for row in reader:
    # 在reader中循环，要遍历所有的对象(文件的所有的行），找出每一行的下标4的最高值TMAX
    high = int(row[4])
    # 必须有next(reader)，才能把第一行拿出来并且把第一行改成能够访问的列表形式，不然row[4]不知到是从哪里找。
    # 没有next(reader),row[4]是CSV对象（一个迭代器）的第五个元素TMAX,而字符串不能int()。

    # int() 只接受看起来像整数的字符串。也就是'56','44'这样的，才能int()后，变成56，44.
    # 使用int()的必要性：后续可以计算。如果不用int()，得到的是'56','77',这样的字符串。计算会出错。
    highs.append(high)

print(highs)
# 因为你的 reader 在进入 for 循环之前，已经被 next(reader) 消耗掉了一行。
# 所以打印结果不会出现第一行的下标4的TMAX
```

```python
# 添加最低温度
# 绘制全年高温时间表
from datetime import datetime
# 从datetime模块导入datetime类

import matplotlib.pyplot as plt
from pathlib import Path
import csv

path = Path("weather_data/death_valley_2021_simple.csv")
lines = path.read_text().splitlines()

reader = csv.reader(lines)
header_row = next(reader)

print(list(enumerate(header_row)))

for index, column_header in enumerate(header_row):
    print(index, column_header)


dates, highs, lows = [], [], []
for row in reader:
    current_date = datetime.strptime(row[2], "%Y-%m-%d")
    # 从reader，csv.reader读出内容
    try:
        high = int(row[3])
        low = int(row[4])
    except ValueError:
        print(f"Missing data for {current_date}")
    else:
        dates.append(current_date)
        highs.append(high)
        lows.append(low)

# 总结帮你彻底分清：
# 代码结构	                        如果发生异常，程序流会怎样？	                要不要写 continue？
# try/except 后面直接跟 append	    会继续向下执行 append	                        必须写 continue，否则坏数据会进列表
# try/except 后面跟 else	        会跳过 else，直接进入下一次 for 循环	        不需要写 continue，因为 else 已经帮你挡住了
# 💡 核心一句：
# 没有 else，continue 就是「安全阀门」；有 else，else 本身就是那个「安全阀门」。


fig, ax = plt.subplots()
ax.plot(dates, highs, linewidth=1, color="red", alpha=1)
# 实参alpha=0.5: 指定color='red'的透明度。
ax.plot(dates, lows, linewidth=1, color="blue", alpha=0.5)
# 用matplotlib做图，要在一个画面fig里的ax里画两条线，需要ax.plot()两次。

ax.fill_between(dates, highs, lows, facecolor="green", alpha=0.2)
# 方法fill_between():dates是x轴，highs,lows是y轴的两个值，之间的颜色facecolor='blue',透明度 alpha=0.5


ax.set_title("Daily high temperatures, 2021 of death_valley", fontsize=14)
ax.set_xlabel("days", fontsize=14)
fig.autofmt_xdate()
# 自动格式一下xz轴上的日期，防止重叠。只有当你 X 轴的数据是真正的“时间对象”（datetime）时，autofmt_xdate 才会生效。
# autofmt_xdate() 是作用在“整张画布（fig）”上的全局排版操作，而不是作用在“某个具体的坐标轴（ax）”上的绘图操作，所以用fig调用而非ax调用
ax.set_ylabel("temperature", fontsize=14)
ax.tick_params(labelsize=12)

plt.show()
```

```python
# 练习16.2 比较锡特卡和死亡谷的温度
# 在有关锡特卡和死亡谷的图中，温度刻度表示的数据范围不同。为了准确的比较锡特卡和死亡谷的温度范围，需要在y轴上使用相同的刻度。
# 为此，要修改y轴的设置，对锡特卡和死亡谷的温度范围进行直接比较。（也可也对任意两个地方的温度范围进行比较。）

import csv
from pathlib import Path
import matplotlib.pyplot as plt
from datetime import datetime

path_sitka = Path("weather_data/sitka_weather_2021_full.csv")
path_death = Path("weather_data/death_valley_2021_full.csv")

lines_sitka = path_sitka.read_text().splitlines()
lines_death = path_death.read_text().splitlines()

reader_sitka = csv.reader(lines_sitka)
reader_death = csv.reader(lines_death)

header_row_sitka = next(reader_sitka)
header_row_death = next(reader_death)

# print(header_row_sitka)
# print(header_row_death)

for index, column_sitka in enumerate(header_row_sitka):
    print(index, column_sitka)
for index, column_death in enumerate(header_row_death):
    print(f"\n{index} {column_death}")

for i in range(len(header_row_sitka)):
    if header_row_sitka[i] == "TMAX":
        print(i)

for i in range(len(header_row_death)):
    if header_row_death[i] == "TMAX":
        print(i)
# print(header_row_death.index("TMAX"))

dates_sitka = []
dates_death = []
TMAXs_sitka = []
TMAXs_death = []

for sitka_row in reader_sitka:
    try:
        current_date_sitka = datetime.strptime(sitka_row[2], "%Y-%m-%d")
        TMAX_sitka = float(sitka_row[7])
    except ValueError:
        print("data missing")
    else:
        dates_sitka.append(current_date_sitka)
        TMAXs_sitka.append(TMAX_sitka)

for death_row in reader_death:
    try:
        current_date_death = datetime.strptime(death_row[2], "%Y-%m-%d")
        TMAX_death = float(death_row[6])
    except ValueError:
        print("data missing")
    else:
        dates_death.append(current_date_death)
        TMAXs_death.append(TMAX_death)

# if len(dates_sitka) >= len(dates_death):
#     dates = dates_sitka
# else:
#     dates = dates_death

fig, ax = plt.subplots(figsize=(15, 6))

ax.plot(dates_sitka, TMAXs_sitka, linewidth=1, color="blue", label="sitka temperature")
ax.plot(
    dates_death, TMAXs_death, linewidth=1, color="red", label="death_valley temperature"
)

# ✅ 强制让 Y 轴刻度一致（核心要求）
# 找出两个数据集里最大的温度和最小的温度，统一设定
min_temp = min(min(TMAXs_sitka), min(TMAXs_death))
max_temp = max(max(TMAXs_sitka), max(TMAXs_death))
ax.set_ylim(min_temp - 5, max_temp + 5)
# 上下留 5 度的余地，视觉更好
# 如果不强制设置 ax.set_ylim()，Matplotlib 会在画第一条线时，根据锡特卡的数据自动把 Y 轴设为 40~70。
# 然后在画第二条线时，它会发现死亡谷的数据远远超出了这个范围，于是它会自动拉长 Y 轴到 50~120。

ax.set_title("Daily High Temperatures: Sitka vs Death Valley", fontsize=16)
ax.set_xlabel("Date", fontsize=14)
ax.set_ylabel("Temperature (F)", fontsize=14)
ax.legend()
#  ax.legend() 会自动根据你画线时传入的 label 参数生成图例。
# 为了让它生效，你在画线时必须给每一组数据加上 label 参数。
fig.autofmt_xdate()

ax.fill_between(
    TMAXs_sitka,
    TMAXs_death,
    color="green",
)

plt.show()
```
