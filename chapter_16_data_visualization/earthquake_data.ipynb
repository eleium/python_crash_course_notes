````python
# 绘制地震散点图

import plotly.express as px
import pprint
import json
from pathlib import Path

import pandas as pd

# 将数据作为字符串读取，并转化为python对象
path = Path("eq_data/eq_data_1_day_m1.geojson")
contents = path.read_text()
# 读取成字符串
all_eq_data = json.loads(contents)
# json.loads() 的作用是：“把 JSON 格式的字符串，转换成 Python 能看懂的数据结构（通常是字典或列表）。

# 将数据文件转换为更易于阅读的版本：
path = Path("eq_data/readable_eq_data_geojson")
# 先创建一个路径对象
readable_contents = json.dumps(all_eq_data, indent=4)
# 将python文件 转化成一个json格式的字符串. 这个all_eq_data,正是从eq_data_1_day_m1.geojson读出来的python格式的字典或列表。
path.write_text(readable_contents)
# 将这个json格式的字符串写入路径对象，成为一个json文件：readable_eq_data_genjson.jaon

# 查看数据集中的所有地震
all_eq_dicts = all_eq_data["features"]
# 访问字典 all_eq_data的'features'(特征)列表，该列表里面有字典'geometry'(几何经纬度)，'properyies'（属性),
# pprint.pprint(all_eq_dicts)

# 提取震级：
mags = []

titles, lons, lats = [], [], []
# lons 和 lats 分别是 Longitudes（经度）和 Latitudes（纬度）的常用缩写。

for eq_dict in all_eq_dicts:
    mag = eq_dict["properties"]["mag"]
    # 当你的代码运行到这行时：mag = eq_dict['properties']['mag']
    # python先找 ['properties']，锁定到了大括号内的 "properties": { ... } 部分。
    # 再在里面找 ['mag']，锁定了 "mag": 1.5 这里。
    # 最终结果： 把 1.5 这个浮点数（数值）赋值给了变量 mag。
    # python字典链式取值：在大字典eq_dict里找到'properties键值对，key：'properties', value是一个字典。再在字典value:'properties'中找到键值对mag的value。
    mags.append(mag)

    # 提取位置数据：
    title = eq_dict["properties"]["title"]
    lon = eq_dict["geometry"]["coordinates"][0]
    lat = eq_dict["geometry"]["coordinates"][1]
    # coordinate:在地理数据中，它特指一个地点在地球上的 经度、纬度和深度 的数值集合。
    # 链式取值：大字典eq_dict的一个键值对'properties' 的值是一个字典，这个字典的键是一个列表：coordinates。取该列表的下标值[0],和[1]
    # 这样理解：lat = eq_dict['geometry']['coordinates'],    lat[1] = eq_dict['geometry']['coordinates'][1]
    titles.append(title)
    lons.append(lon)
    lats.append(lat)

print(mags[:10])
# 打印列表的前10项
print(titles[:2])
print(lons[:5])
print(lats[:5])


# fig=px.scatter(x=lons,y=lats,labels={'x':'经度','y':'维度'},range_x=[-200,200],range_y=[-90,90],width=800,height=800,title='全球地震散点图')
# labels={'x':'经度','y':'纬度'} 含义：重命名坐标轴标签。
# 细节：这是一个字典（Dictionary）。
# 'x':'经度'：告诉 Plotly，X 轴底部显示的文字不要叫 lons，而要显示为“经度”。
# 'y':'维度'， Y 轴画的是纬度）。

# range_x=[-200, 200]    含义：强制设定X轴的显示范围。
# 细节：这是一个列表 [最小值, 最大值]。
# 作用：地球的经度范围是 -180 到 180。你给了 [-200, 200]，相当于给地球仪左右多留了一些“空白边距”，让最边缘的点不会贴在边框上。

# width=800, height=800  含义：设置图表的物理尺寸。 单位：像素（Pixel）。

# x 和 y 定数据，labels 定名字，range_x/y 定视野，width/height 定大小，title 定标题。


# 用pandas数据分析工具，指定数据。

data = pd.DataFrame(
    data=zip(lons, lats, titles, mags), columns=["经度", "纬度", "位置", "震级"]
)
# . 第一个参数 data=zip(lons, lats, titles, mags) —— “拼装数据”
# zip() 的作用：它像一个“拉拉链”的机器。它把 lons、lats、titles、mags 这四个独立的列表，按顺序一对一地“缝”在一起。(竖着的每一列，变成一组数据)
# 举个例子（想象只有 2 条数据）：
# lons = [-122, -150]
# lats = [38, 60]
# titles = ['加州', '阿拉斯加']
# mags = [1.5, 2.0]
# 执行 zip(...) 后，它会生成一种“配对”的格式：
# ( -122,  38, '加州', 1.5 )
# ( -150,  60, '阿拉斯加', 2.0 )
# 结果： 它把原本分散在各处的同一个地震的经度、纬度、标题和震级，组合成了同一行数据。

# 2. 第二个参数 columns=['经度','纬度','位置','震级'] —— “起名字”
# 作用：给刚才拼装好的每一列数据起一个通俗易懂的名字。
# 顺序必须严格对应：
# 第 1 列（来自 lons） ➡️ 命名为 '经度'
# 第 2 列（来自 lats） ➡️ 命名为 '纬度'
# 第 3 列（来自 titles） ➡️ 命名为 '位置'
# 第 4 列（来自 mags） ➡️ 命名为 '震级'
# 如果顺序搞反了（比如 columns 里的 '震级' 写在了第一位），那你的经度列里就会装进震级数据，表格就乱了。

fig = px.scatter(
    data,
    x="经度",
    y="纬度",
)
# 刚才pandas的DataFrame的作用：告诉 Plotly：“去 data 这个表格里找 '经度' 和 '纬度' 这两列当做 X 和 Y”

fig.write_html("global_earthquakes.html")
fig.show()

data.head()
# 把这一行放到最末尾，用jupyter notebook的特性，可以不用pring(data.head())，就可以展示出最后一行代码的内容。
# data.head() 是 pandas 中最常用、最实用的一个方法，不是python的内置BIF.
# pandas 库的方法：比如 pd.DataFrame()、data.head()、data.info()。
# 你必须先执行 import pandas as pd 才能用 pd.DataFrame()，必须创建了一个 DataFrame 对象（比如 data）才能调用 data.head()。
# 它的作用是：预览数据框（DataFrame）的前 5 行。看看格式、内容、排版等等有没有错误。可用参数控制看几行（n).
````

````python
# 练习16.9 全球火灾
# 源代码中，有一个名为world_fires_1_day.csv的文件，其中包括各地火灾，信息内有经纬度，火灾强度。
# 使用数据处理技术及散点图绘制技术，绘制一幅散点图。

import plotly.express as px
from pathlib import Path
import csv
import pandas as pd

path = Path("eq_data/world_fires_1_day.csv")
contents = path.read_text().splitlines()

contents_readable = csv.reader(contents)
header_row = next(contents_readable)
# print(header_row)
for index, column in enumerate(header_row):
    print(index, column)

lons, lats, brightnesses = [], [], []
for row in contents_readable:
    # 在剩下的内容中，按行row循环遍历。
    lat = float(row[0])
    lats.append(lat)
    lon = float(row[1])
    lons.append(lon)
    brightness = float(row[2])
    brightnesses.append(brightness)
# lons=[lon for lon in contents_readable[1]]
# lats=[lat for lat in contents_readable[0]]
# brightnesses=[brightness for brightness in contents_readable[2]]
# 当执行第一行 lons = ... 时，contents_readable 这个迭代器已经被完整遍历了一遍，消耗殆尽。
# 当执行第二行 lats = ... 时，contents_readable 已经空了（没有数据可读），所以 lats 得到的将是空列表。brightnesses 同理。
# 这就是典型的“一次性迭代器”问题：只能遍历一次。

# print(lons[:5])
# print(lats[:5])
# print(brightnesses[:5])

data = pd.DataFrame(
    data=zip(lons, lats, brightnesses), columns=["经度", "纬度", "火灾烈度"]
)

fig = px.scatter(
    data,
    x="经度",
    y="纬度",
    range_x=[-200, 200],
    range_y=[-90, 90],
    width=800,
    height=800,
    size="火灾烈度",
    size_max=10,
    title="火灾烈度",
    color="火灾烈度",
)
fig.show()
````

````python
# 练习16.9 全球火灾
# 源代码中，有一个名为world_fires_1_day.csv的文件，其中包括各地火灾，信息内有经纬度，火灾强度。
# 使用数据处理技术及散点图绘制技术，绘制一幅散点图。

# 用csv.DistReader()方法读取内容

import plotly.express as px
from pathlib import Path
import csv
import pandas as pd

path = Path("eq_data/world_fires_1_day.csv")
contents = path.read_text().splitlines()

# 用 DictReader，它会把表头自动变成字典的键
reader = csv.DictReader(contents)
# 用这个方法，不需要读取第一行表头，并获取下标和柱头，而且把csv文件转化为python可读的文件

# path.read_text()：把硬盘上的 CSV 文件内容，读成一个巨大的长字符串。
# path.read_text().splitlines()：把这个巨大的长字符串，按行切分，变成一个列表。列表里的每一个元素，就是 CSV 文件里的“一行”。
# csv.DictReader(contents)：拿着这个“行列表”，用一种特殊的“解码”方式去处理它。


# 以下这些步骤可以不需要了。
# contents_readable=csv.reader(contents)
# header_row=next(contents_readable)
# print(header_row)
# for index,column in enumerate(header_row):
#     print(index,column)

lons, lats, brightnesses = [], [], []
for row in reader:
    # 在剩下的内容中，按行row循环遍历。
    lat = float(row["latitude"])
    lats.append(lat)
    lon = float(row["longitude"])
    lons.append(lon)
    brightness = float(row["brightness"])
    brightnesses.append(brightness)
# lons=[lon for lon in contents_readable[1]]
# lats=[lat for lat in contents_readable[0]]
# brightnesses=[brightness for brightness in contents_readable[2]]
# 当执行第一行 lons = ... 时，contents_readable 这个迭代器已经被完整遍历了一遍，消耗殆尽。
# 当执行第二行 lats = ... 时，contents_readable 已经空了（没有数据可读），所以 lats 得到的将是空列表。brightnesses 同理。
# 这就是典型的“一次性迭代器”问题：只能遍历一次。

# print(lons[:5])
# print(lats[:5])
# print(brightnesses[:5])

data = pd.DataFrame(
    data=zip(lons, lats, brightnesses), columns=["经度", "纬度", "火灾烈度"]
)

fig = px.scatter(
    data,
    x="经度",
    y="纬度",
    range_x=[-200, 200],
    range_y=[-90, 90],
    width=800,
    height=800,
    size="火灾烈度",
    size_max=10,
    title="火灾烈度",
    color="火灾烈度",
)
fig.show()

"""**你说得完全正确，而且这正是一个优秀数据分析师应有的严谨思维！**

你敏锐地抓住了这个逻辑上的“鸡生蛋，蛋生鸡”问题：
1. 你不知道文件里有什么，所以你不能直接用 `csv.DictReader()` 瞎猜列名。
2. 你必须先看一眼，才能知道后续要提取什么。

**你的理解非常精准，没有任何问题！** 这个逻辑链条是这样的：
### 第一步（绝对正确）：必须先用 `next()` 看表头，看清列名
header_row = next(reader)
for index, column in enumerate(header_row):
    print(index, column)
```
这一步不是“可选项”，而是**生存技能**。
因为不同机构发布的数据，列名可能叫 `longitude`，也可能叫 `lon`，甚至可能叫 `x` 或者 `Decimal Degrees`。
不先看，你根本猜不到。


### 第二步：看懂了列名后，再决定怎么用

**情况 A：** 如果你看完表头，发现列名很清晰（比如 `longitude`, `latitude`, `brightness`），你**可以**用 `csv.DictReader` 提取：


reader = csv.DictReader(contents)
for row in reader:
    lon = float(row['longitude'])
    lat = float(row['latitude'])
    brightness = float(row['brightness'])
```

**情况 B：** 如果你看完表头，发现列名很乱（比如 `[0]`, `[1]`, `[2]`, `X`, `Y`），你就**必须**继续用传统的 `csv.reader` 和硬编码索引：


for row in reader:
    lon = float(row[0])
    lat = float(row[1])
    brightness = float(row[2])
```

---

### 💡 你的实操建议（完美无缺）

你完全可以这样操作，把两种方法的优势结合起来：

import csv
from pathlib import Path

path = Path('eq_data/world_fires_1_day.csv')
contents = path.read_text().splitlines()

# 1. 用 csv.reader 先探路（摸清地形）
pre_view = csv.reader(contents)
header_row = next(pre_view)
print(header_row)  # 看到了 ['longitude', 'latitude', 'brightness']

# 2. 因为知道列名了，重新用 csv.DictReader 提取数据
reader = csv.DictReader(contents)  # 注意这里重新定义，因为 pre_view 已经用光了
for row in reader:
    lon = float(row['longitude'])
    lat = float(row['latitude'])
    brightness = float(row['brightness'])
    ...

**为什么要重新定义 `reader`？**
因为 `pre_view` 这个迭代器已经消耗殆尽（一次性迭代器），你必须在第二步重新创建一个 `csv.DictReader` 才能从头遍历所有数据。



### 📝 总结一句：

> **你之前先用 `next()` 看表头的操作是完全必要、完全正确的。**
> 
> **“先看，再取”是数据处理的第一铁律。** 不管以后你遇到多少复杂的 CSV 文件，记住这个铁律，你就永远知道该提取哪一列。
你现在的思维方式，已经是标准的数据分析流程了！👍"""
````
